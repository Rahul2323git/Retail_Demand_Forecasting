```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
PROJECT_ROOT = Path.cwd().parent
RAW_PATH = PROJECT_ROOT / "data" / "raw"

customers = pd.read_csv(RAW_PATH / "customers.csv")
invoice_items = pd.read_csv(RAW_PATH / "invoice_items.csv")
products = pd.read_csv(RAW_PATH / "products.csv")
purchases = pd.read_csv(RAW_PATH / "purchases.csv")
```

```python
datasets = {
    "customers": customers,
    "invoice_items": invoice_items,
    "products": products,
    "purchases": purchases
}

for name, data in datasets.items():
    print(f"\n{'=' * 50}")
    print(name)
    print("Shape:", data.shape)
    print("Columns:", data.columns.tolist())
    print("Duplicates:", data.duplicated().sum())
    print("Missing values:", data.isnull().sum().sum())
```

```python
print("Customers:")
print("CustomerID unique:", customers["CustomerID"].nunique())
print("CustomerID duplicates:", customers["CustomerID"].duplicated().sum())

print("\nProducts:")
print("product_id unique:", products["product_id"].nunique())
print("product_id duplicates:", products["product_id"].duplicated().sum())

print("\nPurchases:")
print("InvoiceID unique:", purchases["InvoiceID"].nunique())
print("InvoiceID + product_id duplicates:",
      purchases[["InvoiceID", "product_id"]].duplicated().sum())

print("\nInvoice Items:")
print("InvoiceID unique:", invoice_items["InvoiceID"].nunique())
print("InvoiceID + product_id duplicates:",
      invoice_items[["InvoiceID", "product_id"]].duplicated().sum())
```

```python
purchases["date"] = pd.to_datetime(
    purchases["date"],
    errors="coerce"
)

print("Invalid dates:", purchases["date"].isna().sum())
print("First date:", purchases["date"].min())
print("Last date:", purchases["date"].max())
print("Unique dates:", purchases["date"].nunique())
```

```python
print("Purchase quantity < 0:")
print((purchases["quantity"] < 0).sum())

print("\nInvoice quantity < 0:")
print((invoice_items["quantity"] < 0).sum())

print("\nPrice < 0:")
print((invoice_items["price"] < 0).sum())

print("\nLine total < 0:")
print((invoice_items["line_total"] < 0).sum())
```

```python
print("Top quantities:")
print(
    purchases
    .sort_values("quantity", ascending=False)
    [["InvoiceID", "date", "CustomerID", "product_id", "quantity"]]
    .head(20)
)
```

```python
print("Top prices:")
print(
    invoice_items
    .sort_values("price", ascending=False)
    [["InvoiceID", "product_id", "quantity", "price", "line_total"]]
    .head(20)
)
```

```python
print("Top line totals:")
print(
    invoice_items
    .sort_values("line_total", ascending=False)
    [["InvoiceID", "product_id", "quantity", "price", "line_total"]]
    .head(20)
)
```

```python
invoice_items["calculated_total"] = (
    invoice_items["quantity"] * invoice_items["price"]
)

print(
    "Line total mismatches:",
    (
        ~np.isclose(
            invoice_items["line_total"],
            invoice_items["calculated_total"],
            rtol=0.01
        )
    ).sum()
)
```

```python
purchase_customers = set(purchases["CustomerID"])
customer_ids = set(customers["CustomerID"])

missing_customers = purchase_customers - customer_ids

print("CustomerIDs in purchases:", len(purchase_customers))
print("CustomerIDs in customers:", len(customer_ids))
print("Missing CustomerIDs:", len(missing_customers))
```

```python
purchase_products = set(purchases["product_id"])
product_ids = set(products["product_id"])

missing_products_purchase = purchase_products - product_ids

print("ProductIDs in purchases:", len(purchase_products))
print("ProductIDs in products:", len(product_ids))
print("Missing ProductIDs:", len(missing_products_purchase))
```

```python
invoice_products = set(invoice_items["product_id"])

missing_products_invoice = invoice_products - product_ids

print(
    "Missing ProductIDs in invoice_items:",
    len(missing_products_invoice)
)
```

```python
purchase_invoices = set(purchases["InvoiceID"])
invoice_item_invoices = set(invoice_items["InvoiceID"])

missing_invoices = invoice_item_invoices - purchase_invoices

print("Invoices in purchases:", len(purchase_invoices))
print("Invoices in invoice_items:", len(invoice_item_invoices))
print("Missing invoices:", len(missing_invoices))
```

```python
purchase_qty = (
    purchases
    .groupby(["InvoiceID", "product_id"])["quantity"]
    .sum()
    .reset_index(name="purchase_quantity")
)

invoice_qty = (
    invoice_items
    .groupby(["InvoiceID", "product_id"])["quantity"]
    .sum()
    .reset_index(name="invoice_quantity")
)
```

```python
quantity_check = purchase_qty.merge(
    invoice_qty,
    on=["InvoiceID", "product_id"],
    how="outer"
)
```

```python
print(
    "Missing in purchases:",
    quantity_check["purchase_quantity"].isna().sum()
)

print(
    "Missing in invoice_items:",
    quantity_check["invoice_quantity"].isna().sum()
)
```

```python
purchase_invoices = set(purchases["InvoiceID"])
invoice_item_invoices = set(invoice_items["InvoiceID"])

missing_invoices = invoice_item_invoices - purchase_invoices

print("Invoices in purchases:", len(purchase_invoices))
print("Invoices in invoice_items:", len(invoice_item_invoices))
print("Missing invoices:", len(missing_invoices))
```

```python
purchase_qty = (
    purchases
    .groupby(["InvoiceID", "product_id"])["quantity"]
    .sum()
    .reset_index(name="purchase_quantity")
)

invoice_qty = (
    invoice_items
    .groupby(["InvoiceID", "product_id"])["quantity"]
    .sum()
    .reset_index(name="invoice_quantity")
)
```

```python
quantity_check = purchase_qty.merge(
    invoice_qty,
    on=["InvoiceID", "product_id"],
    how="outer"
)
```

```python
print(
    "Missing in purchases:",
    quantity_check["purchase_quantity"].isna().sum()
)

print(
    "Missing in invoice_items:",
    quantity_check["invoice_quantity"].isna().sum()
)
```

```python
quantity_check["quantity_match"] = np.isclose(
    quantity_check["purchase_quantity"].fillna(0),
    quantity_check["invoice_quantity"].fillna(0)
)

print(
    "Quantity mismatches:",
    (~quantity_check["quantity_match"]).sum()
)
```

```python
product_consistency = (
    products
    .groupby("product_id")
    .agg({
        "item": "nunique",
        "category": "nunique",
        "price": "nunique"
    })
)

print(product_consistency.max())
```
