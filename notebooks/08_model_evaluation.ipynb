```python
# ============================================================
# STEP 11.1 — CREATE MODEL EVALUATION SETUP
# ============================================================

from pathlib import Path
import json
from datetime import datetime

# ------------------------------------------------------------
# Project paths
# ------------------------------------------------------------

PROJECT_ROOT = Path(r"D:\All ML Projects\Retail_Demand_Forecasting")

MODELS_DIR = PROJECT_ROOT / "data" / "models"
TUNING_DIR = MODELS_DIR / "hyperparameter_tuning"

# ------------------------------------------------------------
# Final model artifact paths
# ------------------------------------------------------------

HGB_MODEL_PATH = TUNING_DIR / "hgb_tuned_model.joblib"
CATBOOST_MODEL_PATH = TUNING_DIR / "catboost_tuned_model.cbm"

# ------------------------------------------------------------
# Step 11 evaluation output directory
# ------------------------------------------------------------

# Reuse the existing Step 10 model directory.
# No new folder is being created.
TUNING_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# Final model roles
# ------------------------------------------------------------

PRIMARY_MODEL = "HistGradientBoosting"
SECONDARY_MODEL = "CatBoost"

# ------------------------------------------------------------
# Evaluation metrics
# ------------------------------------------------------------

PRIMARY_METRICS = ["MAE", "RMSE"]
SECONDARY_METRICS = ["sMAPE"]
DIAGNOSTIC_METRICS = ["MAPE"]

ALL_METRICS = PRIMARY_METRICS + SECONDARY_METRICS + DIAGNOSTIC_METRICS

# ------------------------------------------------------------
# Test-set policy
# ------------------------------------------------------------

TEST_SET_ALLOWED_FROM_STEP = 11

# ------------------------------------------------------------
# Setup timestamp
# ------------------------------------------------------------

SETUP_TIMESTAMP = datetime.now().isoformat(timespec="seconds")

# ------------------------------------------------------------
# Display setup
# ------------------------------------------------------------

print("=" * 70)
print("STEP 11 — MODEL EVALUATION")
print("11.1 — EVALUATION SETUP")
print("=" * 70)

print(f"\nProject root:")
print(PROJECT_ROOT)

print(f"\nEvaluation output directory:")
print(TUNING_DIR)

print("\nFinal models:")
print(f"  Primary   : {PRIMARY_MODEL}")
print(f"  Secondary : {SECONDARY_MODEL}")

print("\nModel artifacts:")
print(f"  HGB       : {HGB_MODEL_PATH}")
print(f"  CatBoost  : {CATBOOST_MODEL_PATH}")

print("\nMetrics:")
print(f"  Primary   : {PRIMARY_METRICS}")
print(f"  Secondary : {SECONDARY_METRICS}")
print(f"  Diagnostic: {DIAGNOSTIC_METRICS}")

print("\nTest-set policy:")
print("  Test set will ONLY be used for FINAL TEST EVALUATION in Step 11.")

print("\nSetup timestamp:")
print(SETUP_TIMESTAMP)

print("\n" + "=" * 70)
print("STEP 11.1 SETUP COMPLETE")
print("No test-set data was loaded or used.")
print("=" * 70)
```

```python
# ============================================================
# STEP 11.2 — VERIFY FINAL MODELS AND UNTOUCHED TEST SET
# ============================================================

import pandas as pd
import joblib
from catboost import CatBoostRegressor

# ------------------------------------------------------------
# Test-set path
# ------------------------------------------------------------

TEST_PATH = PROJECT_ROOT / "data" / "processed" / "test.csv"

print("=" * 70)
print("STEP 11 — MODEL EVALUATION")
print("11.2 — VERIFY FINAL MODELS AND UNTOUCHED TEST SET")
print("=" * 70)

# ============================================================
# 1. VERIFY MODEL ARTIFACT FILES
# ============================================================

print("\n[1] VERIFYING FINAL MODEL ARTIFACTS")
print("-" * 70)

print(f"HGB model exists      : {HGB_MODEL_PATH.exists()}")
print(f"CatBoost model exists : {CATBOOST_MODEL_PATH.exists()}")

if not HGB_MODEL_PATH.exists():
    raise FileNotFoundError(
        f"HistGradientBoosting model not found:\n{HGB_MODEL_PATH}"
    )

if not CATBOOST_MODEL_PATH.exists():
    raise FileNotFoundError(
        f"CatBoost model not found:\n{CATBOOST_MODEL_PATH}"
    )

# ============================================================
# 2. LOAD FINAL MODELS
# ============================================================

print("\n[2] LOADING FINAL MODELS")
print("-" * 70)

hgb_model = joblib.load(HGB_MODEL_PATH)

catboost_model = CatBoostRegressor()
catboost_model.load_model(str(CATBOOST_MODEL_PATH))

print("HistGradientBoosting model loaded successfully.")
print("CatBoost model loaded successfully.")

# ============================================================
# 3. VERIFY TEST FILE
# ============================================================

print("\n[3] VERIFYING TEST SET")
print("-" * 70)

print(f"Test file exists: {TEST_PATH.exists()}")

if not TEST_PATH.exists():
    raise FileNotFoundError(
        f"Test set not found:\n{TEST_PATH}"
    )

# Load test data for FINAL TEST EVALUATION preparation.
test_df = pd.read_csv(TEST_PATH)

print(f"Test shape: {test_df.shape}")

# ============================================================
# 4. VERIFY EXPECTED TEST ROW COUNT
# ============================================================

EXPECTED_TEST_ROWS = 81521

print("\n[4] VERIFYING TEST ROW COUNT")
print("-" * 70)

print(f"Expected rows: {EXPECTED_TEST_ROWS}")
print(f"Actual rows  : {len(test_df)}")

if len(test_df) == EXPECTED_TEST_ROWS:
    print("Row-count verification: PASSED")
else:
    print("Row-count verification: WARNING")
    print("The actual row count differs from the Step 7 expected count.")

# ============================================================
# 5. DISPLAY TEST COLUMNS
# ============================================================

print("\n[5] TEST SET COLUMNS")
print("-" * 70)

print(f"Number of columns: {len(test_df.columns)}")

for i, column in enumerate(test_df.columns, start=1):
    print(f"{i:2d}. {column}")

# ============================================================
# 6. CHECK FOR MISSING VALUES
# ============================================================

print("\n[6] BASIC TEST SET STRUCTURE CHECK")
print("-" * 70)

missing_total = test_df.isna().sum().sum()
duplicate_rows = test_df.duplicated().sum()

print(f"Total missing values : {missing_total}")
print(f"Duplicate rows       : {duplicate_rows}")

# ============================================================
# 7. IDENTIFY DATE COLUMN
# ============================================================

print("\n[7] VERIFYING TEST DATE RANGE")
print("-" * 70)

possible_date_columns = [
    column for column in test_df.columns
    if "date" in column.lower()
]

print(f"Possible date columns: {possible_date_columns}")

if possible_date_columns:
    DATE_COLUMN = possible_date_columns[0]

    test_df[DATE_COLUMN] = pd.to_datetime(
        test_df[DATE_COLUMN],
        errors="coerce"
    )

    print(f"Date column used for verification: {DATE_COLUMN}")
    print(f"Minimum test date: {test_df[DATE_COLUMN].min()}")
    print(f"Maximum test date: {test_df[DATE_COLUMN].max()}")

    EXPECTED_TEST_START = pd.Timestamp("2015-09-12")
    EXPECTED_TEST_END = pd.Timestamp("2015-12-30")

    if (
        test_df[DATE_COLUMN].min() == EXPECTED_TEST_START
        and
        test_df[DATE_COLUMN].max() == EXPECTED_TEST_END
    ):
        print("Date-range verification: PASSED")
    else:
        print("Date-range verification: WARNING")
        print("Actual dates differ from the expected Step 7 test period.")

else:
    DATE_COLUMN = None
    print("No obvious date column was automatically identified.")

# ============================================================
# 8. MODEL TYPE VERIFICATION
# ============================================================

print("\n[8] MODEL TYPE VERIFICATION")
print("-" * 70)

print(f"HGB model type      : {type(hgb_model).__name__}")
print(f"CatBoost model type : {type(catboost_model).__name__}")

# ============================================================
# FINAL STATUS
# ============================================================

print("\n" + "=" * 70)
print("STEP 11.2 VERIFICATION COMPLETE")
print("=" * 70)

print("\nIMPORTANT:")
print("- Final models were loaded successfully.")
print("- Test set was accessed for FINAL TEST EVALUATION.")
print("- No predictions were generated.")
print("- No evaluation metrics were calculated.")
print("- No model selection or tuning was performed.")
print("=" * 70)
```
